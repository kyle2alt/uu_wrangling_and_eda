```python
import numpy as np
import pandas as pd

df = pd.read_csv("ForeignGifts_edu.csv")
df.head()
```

```python
print(df["Gift Type"].value_counts())
```

```python
print(df.groupby("Gift Type")["Foreign Gift Amount"].describe())
```

The three types of gifts are Contract, Monetary Gift, and Real Estate. Contract is the most common type of gift at roughly 17000, versus Monetary Gift at 10000 and Real Estate at only 11. Contracts also had a larger median at $110,000 versus Monetary Gifts with a median of $50,000. Real Estate gifts were rare but very large, with a median around $668,000. 

Q1: Which Country Gives the most money in total?

```python
df.groupby("Country of Giftor")["Foreign Gift Amount"].sum().nlargest(10)
```

Answer: Qatar

Q2: Which country initiates the most gifts, by count?


```python
df["Country of Giftor"].value_counts().head(10)
```

Answer: England

Q3: Which country gives the largest gifts, on average?

```python
df.groupby("Country of Giftor")["Foreign Gift Amount"].mean().nlargest(10)
```

Answer: Bermuda, Ecuador, Qatar, however Ecuador only has a few gifts in total given versus the other top countries.

Q4: Which institution receive the most money, in total?

```python
df.groupby("Institution Name")["Foreign Gift Amount"].sum().nlargest(10)
```

Answer: Carnegie Mellon

Q5: Which institution receive the greatest number of gifts, by count?

```python
df["Institution Name"].value_counts().head(10)
```

Answer: UCLA
